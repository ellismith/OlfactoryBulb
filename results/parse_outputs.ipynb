```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

try:
    import cPickle
except:
    import pickle as cPickle
```

```python

soma_vs_f = '/home/kedoxey/OB_Model/OlfactoryBulb/results/GammaSignature_SetupTime/soma_vs.pkl'

with open(soma_vs_f, 'rb') as f:
    soma_vs = cPickle.load(f)
    soma_vs.sort(key=lambda row: row[0][0:2])
```

```python
gc_cells = []
mc_cells = []
tc_cells = []
i = 0
j = 0
for cell, t, v in soma_vs:
    if 'GC' in cell:
        gc_cells.append(cell)
    elif 'MC' in cell:
        mc_cells.append((i,cell))
        i += 1
    elif 'TC' in cell:
        tc_cells.append((j,cell))
        j += 1
```

```python
print(len(gc_cells))
print(len(mc_cells))
print(len(tc_cells))
```

```python
fig_width = 27
plt.subplots(figsize=(fig_width, len(soma_vs)*0.12))

i = 0
j = 0
for cell, t, v in soma_vs:
    if 'MC' in cell:
        col = 'blue'
    if 'TC' in cell:
        col = 'red'
    if 'GC' in cell:
        col = 'orange'
        if j > 10:
            continue # don't plot GCs
        
    plt.plot(t,np.array(v)+i,col,label=cell)
    # ax[0].plot(t,np.array(v)+i,col,label=cell)
    i += 100
    j += 1
```
