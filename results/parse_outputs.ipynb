```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

try:
    import cPickle
except:
    import pickle as cPickle
```

```python

soma_vs_f = '/home/kedoxey/OB_Model/OlfactoryBulb/results/GammaSignature_Testing/soma_vs.pkl'

with open(soma_vs_f, 'rb') as f:
    soma_vs = cPickle.load(f)
    soma_vs.sort(key=lambda row: row[0][0:2])
```

```python
gc_cells = []
mc_cells = []
tc_cells = []
i = 0
j = 0
k = 0
gc_thresholds = np.zeros(159)
for cell, t, v in soma_vs:
    if 'GC' in cell:
        gc_cells.append((k,cell))
        count = 0
        voltages = np.array(v)
        for voltage in voltages:
            if voltage > 0:
                count += 1
        gc_thresholds[k] = count
        k += 1
    elif 'MC' in cell:
        mc_cells.append((i,cell))
        i += 1
    elif 'TC' in cell:
        tc_cells.append((j,cell))
        j += 1
```
